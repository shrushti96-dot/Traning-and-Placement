```python
emojis = {
    "happy": "😊",
    "sad": "😞",
    "love": "❤️",
    "angry": "😡",
    "surprised": "🫢"
}
```

```python
sentence = input("Enter a sentence: ")
words = sentence.lower().split()

result = []
for word in words:
    if word in emojis:
        result.append(emojis[word])
    else:
        result.append(word)

print(" ".join(result))
```

```python
import matplotlib.pyplot as plt

# 1. Get the name and clean it up
name = input("Enter a name: ")
processed_name = name.lower().replace(" ", "")

# 2. Count frequencies using a simple dictionary and a loop
letter_counts = {}
for letter in processed_name:
    if letter in letter_counts:
        letter_counts[letter] += 1  # Add 1 if we've seen this letter before
    else:
        letter_counts[letter] = 1   # Set to 1 if it's a new letter

# 3. Separate the dictionary into keys (letters) and values (counts)
letters = list(letter_counts.keys())
frequencies = list(letter_counts.values())

# 4. Create the bar graph
plt.bar(letters, frequencies, color='skyblue', edgecolor='black')
plt.title(f"Letter Frequency in: {name}")
plt.xlabel("Letters")
plt.ylabel("Frequency")

plt.show()
```
